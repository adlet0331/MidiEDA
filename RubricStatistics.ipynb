```python
from RubricStatistics import get_mid_files_statistics
import matplotlib.pyplot as plt

pianovam_stats = get_mid_files_statistics("/Users/simhyeongju/AVAPT/data/pianovam/midi")

import numpy as np
indices = np.arange(1, 10)

# 막대 그래프 생성
plt.bar(indices, pianovam_stats)

# 그래프 제목 및 축 레이블 추가
plt.title('PianoVam Difficulty Distribution\nTotal Counts: {}'.format(sum(pianovam_stats)))
plt.xlabel('Index')
plt.ylabel('Counts')

# x축 눈금을 1부터 9까지 정수로 설정
plt.xticks(indices)

# 그래프를 이미지 파일로 저장
plt.savefig('pianovam_bar_chart.png')
```
